```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline

numeric_features_b = X_b.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features_b = X_b.select_dtypes(
    exclude=np.number
).columns.tolist()

numeric_pipeline_b = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline_b = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])

bottleneck_preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline_b, numeric_features_b),
    ("categorical", categorical_pipeline_b, categorical_features_b)
])
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_bottleneck = Pipeline([
    ("preprocessor", bottleneck_preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=500,
        max_depth=12,
        min_samples_leaf=3,
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
rf_bottleneck.fit(
    X_train_b,
    y_train_b
)
```

```python
y_pred_rf_b = rf_bottleneck.predict(X_test_b)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

print("Accuracy:",
      accuracy_score(y_test_b, y_pred_rf_b))

print("Precision:",
      precision_score(y_test_b, y_pred_rf_b))

print("Recall:",
      recall_score(y_test_b, y_pred_rf_b))

print("F1:",
      f1_score(y_test_b, y_pred_rf_b))

print("\nClassification Report:")
print(
    classification_report(
        y_test_b,
        y_pred_rf_b,
        target_names=[
            "Not Bottleneck",
            "Bottleneck"
        ]
    )
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test_b,
        y_pred_rf_b
    )
)
```
