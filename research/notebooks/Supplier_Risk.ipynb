```python
# Make a copy before feature engineering
df_model = df.copy()

eps = 1e-6

# 1. Demand vs production capacity
df_model["Demand_Capacity_Ratio"] = (
    df_model["Forecasted_Demand"] /
    (df_model["Production_Capacity"] + eps)
)

# 2. Material coverage
df_model["Material_Demand_Coverage"] = (
    df_model["Raw_Material_Availability"] /
    (df_model["Forecasted_Demand"] + eps)
)

# 3. Material shortage amount
df_model["Material_Shortage"] = (
    df_model["Forecasted_Demand"] -
    df_model["Raw_Material_Availability"]
)

# 4. Supplier delay relative to lead time
df_model["Delay_LeadTime_Ratio"] = (
    df_model["Delivery_Delay"] /
    (df_model["Lead_Time_Supplier"] + eps)
)

# 5. Delivery time relative to supplier lead time
df_model["Delivery_LeadTime_Ratio"] = (
    df_model["Delivery_Time"] /
    (df_model["Lead_Time_Supplier"] + eps)
)

# 6. Quality risk
df_model["Quality_Risk"] = (
    df_model["Defect_Rate"] *
    (1 - df_model["Reliability_Score"])
)

# 7. Production pressure
df_model["Production_Pressure"] = (
    df_model["Machine_Utilization"] / 100
) * (
    df_model["Demand_Capacity_Ratio"]
)

# 8. Logistics burden
df_model["Logistics_Burden"] = (
    df_model["Distance_to_Market"] *
    df_model["Fuel_Cost_per_km"]
)

# 9. Supplier cost exposure
df_model["Supplier_Cost_Exposure"] = (
    df_model["Supplier_Cost_per_Unit"] *
    df_model["Forecasted_Demand"]
)

print(df_model.shape)
```

```python
X = df_model[supplier_features].copy()

y = df_model["Optimization_Label"].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nTarget distribution:")
print(y.value_counts())
```

```python
print(X.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing:", X_test.shape)

print("\nTraining target:")
print(y_train.value_counts())

print("\nTesting target:")
print(y_test.value_counts())
```

```python
numeric_features = supplier_features

print(len(numeric_features))
print(numeric_features)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
```

```python
preprocessor = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(
        n_estimators=500,
        max_depth=12,
        min_samples_leaf=3,
        class_weight="balanced",
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
rf_model.fit(X_train, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test)

print(y_pred_rf[:20])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

print("Accuracy:",
      accuracy_score(y_test, y_pred_rf))

print("Precision:",
      precision_score(
          y_test,
          y_pred_rf,
          average="weighted"
      ))

print("Recall:",
      recall_score(
          y_test,
          y_pred_rf,
          average="weighted"
      ))

print("F1:",
      f1_score(
          y_test,
          y_pred_rf,
          average="weighted"
      ))
```

```python
print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred_rf,
        labels=["Low", "Medium", "High"]
    )
)
```

```python
print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        y_pred_rf,
        labels=["Low", "Medium", "High"]
    )
)
```

```python
report = classification_report(
    y_test,
    y_pred_rf,
    labels=["Low", "Medium", "High"],
    output_dict=True
)

print(
    "HIGH-risk recall:",
    report["High"]["recall"]
)
```

```python
y_binary = (
    df_model["Optimization_Label"] == "High"
).astype(int)

print(y_binary.value_counts())
```

```python
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X,
    y_binary,
    test_size=0.20,
    random_state=42,
    stratify=y_binary
)

print(y_train_b.value_counts())
print(y_test_b.value_counts())
```

```python
high_count = (y_train_b == 1).sum()
non_high_count = (y_train_b == 0).sum()

scale_pos_weight = non_high_count / high_count

print("scale_pos_weight:", scale_pos_weight)
```

```python
high_risk_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", XGBClassifier(
        n_estimators=400,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=scale_pos_weight,
        eval_metric="logloss",
        random_state=42
    ))
])
```

```python
high_risk_model.fit(
    X_train_b,
    y_train_b
)
```

```python
high_prob = high_risk_model.predict_proba(X_test_b)[:, 1]
```

```python
high_pred = (high_prob >= 0.50).astype(int)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(
    classification_report(
        y_test_b,
        high_pred,
        target_names=["Not High", "High"]
    )
)

print(
    confusion_matrix(
        y_test_b,
        high_pred
    )
)
```
